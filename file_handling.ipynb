## Write a program to simulate an ATM withdrawal. Use `finally` to display a message that the ATM transaction process has ended whether the transaction succeeds or fails.

```python
# ATM withdrawal with a finally block
def atm_withdraw(balance, amount_input):
    try:
        amount = float(amount_input)
        if amount <= 0:
            raise ValueError("Withdrawal amount must be positive.")
        if amount > balance:
            raise ValueError("Insufficient balance.")
        balance -= amount
        print(f"Withdrawal successful. Remaining balance: Rs. {balance:.2f}")
        return balance
    except (ValueError, TypeError) as error:
        print(f"Transaction failed: {error}")
        return balance
    finally:
        print("ATM transaction process has ended.")

atm_withdraw(10000, "2500")
atm_withdraw(10000, "15000")
```

## Create multiple custom exception classes for an online payment system. Handle invalid payment amount, insufficient balance, and unsupported payment method as different business errors.

```python
# Online payment system with separate custom exceptions
class InvalidPaymentAmountError(Exception): pass
class InsufficientBalanceError(Exception): pass
class UnsupportedPaymentMethodError(Exception): pass

def make_payment(amount, method, balance):
    try:
        amount = float(amount)
        if amount <= 0:
            raise InvalidPaymentAmountError("Payment amount must be positive.")
        if method not in {"card", "wallet", "bank"}:
            raise UnsupportedPaymentMethodError(f"Unsupported method: {method}")
        if amount > balance:
            raise InsufficientBalanceError("Not enough balance.")
    except InvalidPaymentAmountError as error:
        print("Invalid amount:", error)
        return balance
    except UnsupportedPaymentMethodError as error:
        print("Payment method error:", error)
        return balance
    except InsufficientBalanceError as error:
        print("Balance error:", error)
        return balance
    else:
        balance -= amount
        print(f"Payment successful via {method}. Remaining balance: Rs. {balance:.2f}")
        return balance

balance = make_payment(1200, "wallet", 5000)
balance = make_payment(7000, "card", balance)
balance = make_payment(500, "crypto", balance)
```

## Create a class-based bank account system. Handle deposits and withdrawals using custom exceptions for invalid transaction amounts and insufficient balance.

```python
# Bank account using custom transaction exceptions
class InvalidTransactionAmountError(Exception): pass
class BankInsufficientBalanceError(Exception): pass

class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = float(balance)

    def deposit(self, amount):
        try:
            amount = float(amount)
            if amount <= 0:
                raise InvalidTransactionAmountError("Deposit must be positive.")
        except (ValueError, TypeError, InvalidTransactionAmountError) as error:
            print("Deposit failed:", error)
            return
        self.balance += amount
        print(f"Deposited Rs. {amount:.2f}; balance: Rs. {self.balance:.2f}")

    def withdraw(self, amount):
        try:
            amount = float(amount)
            if amount <= 0:
                raise InvalidTransactionAmountError("Withdrawal must be positive.")
            if amount > self.balance:
                raise BankInsufficientBalanceError("Insufficient account balance.")
        except (ValueError, TypeError, InvalidTransactionAmountError, BankInsufficientBalanceError) as error:
            print("Withdrawal failed:", error)
            return
        self.balance -= amount
        print(f"Withdrawn Rs. {amount:.2f}; balance: Rs. {self.balance:.2f}")

account = BankAccount("Hassan", 5000)
account.deposit(1500)
account.withdraw(2000)
account.withdraw(10000)
```

## Create an e-commerce order system that handles invalid products, invalid quantities, insufficient stock, and invalid payments. Use custom exceptions for business-rule failures.

```python
# E-commerce order system with custom business exceptions
class InvalidProductError(Exception): pass
class InvalidQuantityError(Exception): pass
class InsufficientStockError(Exception): pass
class InvalidPaymentError(Exception): pass

inventory = {"laptop": {"price": 80000.0, "stock": 3},
             "mouse": {"price": 1200.0, "stock": 10}}

def place_order(product, quantity, payment):
    try:
        if product not in inventory:
            raise InvalidProductError(f"{product!r} is not available.")
        quantity = int(quantity)
        if quantity <= 0:
            raise InvalidQuantityError("Quantity must be positive.")
        if inventory[product]["stock"] < quantity:
            raise InsufficientStockError("Not enough stock.")
        payment = float(payment)
        total = inventory[product]["price"] * quantity
        if payment < total:
            raise InvalidPaymentError(f"Payment must be at least Rs. {total:.2f}.")
    except (InvalidProductError, InvalidQuantityError, InsufficientStockError, InvalidPaymentError, ValueError, TypeError) as error:
        print("Order failed:", error)
        return False
    else:
        # Update stock only after every validation passes.
        inventory[product]["stock"] -= quantity
        print(f"Order successful. Total: Rs. {total:.2f}; stock left: {inventory[product]['stock']}")
        return True

place_order("mouse", 2, 3000)
place_order("laptop", 9, 900000)
```

## ### University Course Registration System

### Goal:
Create a program that allows students to register for university courses.

### Steps:
- Store students and courses.
- Check whether the student exists.
- Check whether the course exists.
- Prevent duplicate registration.
- Check course capacity.
- Check prerequisites.
- Create and handle suitable custom exceptions.

### Example:
Student ID: 102
Course: Python
→ Registration successful

```python
# University course registration system
class StudentNotFoundError(Exception): pass
class CourseNotFoundError(Exception): pass
class DuplicateRegistrationError(Exception): pass
class CourseFullError(Exception): pass
class PrerequisiteError(Exception): pass

students = {
    "102": {"name": "Ali", "completed": ["Intro"]},
    "103": {"name": "Sara", "completed": []}
}
courses = {
    "Python": {"capacity": 2, "registered": [], "prerequisite": "Intro"},
    "Web": {"capacity": 1, "registered": [], "prerequisite": None}
}

def register_student(student_id, course_name):
    try:
        if student_id not in students:
            raise StudentNotFoundError("Student does not exist.")
        if course_name not in courses:
            raise CourseNotFoundError("Course does not exist.")
        student = students[student_id]
        course = courses[course_name]
        if student_id in course["registered"]:
            raise DuplicateRegistrationError("Student is already registered.")
        if len(course["registered"]) >= course["capacity"]:
            raise CourseFullError("Course capacity has been reached.")
        required = course["prerequisite"]
        if required and required not in student["completed"]:
            raise PrerequisiteError(f"Complete {required} first.")
    except (StudentNotFoundError, CourseNotFoundError, DuplicateRegistrationError, CourseFullError, PrerequisiteError) as error:
        print("Registration failed:", error)
        return False
    else:
        course["registered"].append(student_id)
        print(f"Registration successful: {student['name']} -> {course_name}")
        return True

register_student("102", "Python")
register_student("103", "Python")
register_student("102", "Python")
```

## ### Hospital Appointment System

- Store patients and doctors.
- Validate patient and doctor information.

- Prevent duplicate appointment times.
- Create custom exceptions for important business errors.
- Handle invalid input without crashing.

### Example:
Patient ID: P102
Doctor: D05
Time: 10:30 AM
→ Appointment booked successfully

```python
# Hospital appointment system
class PatientNotFoundError(Exception): pass
class DoctorNotFoundError(Exception): pass
class DoctorUnavailableError(Exception): pass
class DuplicateAppointmentError(Exception): pass

patients = {"P102": "Ali", "P103": "Sara"}
doctors = {"D05": {"name": "Dr. Ahmed", "available": ["10:30 AM", "11:30 AM"]}}
appointments = []

def book_appointment(patient_id, doctor_id, time):
    try:
        if patient_id not in patients:
            raise PatientNotFoundError("Patient record not found.")
        if doctor_id not in doctors:
            raise DoctorNotFoundError("Doctor record not found.")
        if time not in doctors[doctor_id]["available"]:
            raise DoctorUnavailableError("Doctor is not available at that time.")
        if any(a["doctor"] == doctor_id and a["time"] == time for a in appointments):
            raise DuplicateAppointmentError("This time is already booked.")
    except (PatientNotFoundError, DoctorNotFoundError, DoctorUnavailableError, DuplicateAppointmentError) as error:
        print("Appointment failed:", error)
        return False
    else:
        appointments.append({"patient": patient_id, "doctor": doctor_id, "time": time})
        print(f"Appointment booked: {patients[patient_id]} with {doctors[doctor_id]['name']} at {time}")
        return True

book_appointment("P102", "D05", "10:30 AM")
book_appointment("P103", "D05", "10:30 AM")
```

## ### Digital Wallet System

### Goal:
Create a class-based digital wallet application.

### Steps:
- Validate wallet ID and PIN.
- Handle inactive accounts.
- Validate transfer amounts.
- Check balance.
- Enforce a transfer limit.
- Create multiple custom exceptions.
- Make sure failed transfers do not change wallet balances.

### Example:
Transfer: Rs. 2,000
→ Transfer successful

```python
# Class-based digital wallet with safe transfers
class WalletError(Exception): pass
class InvalidWalletError(WalletError): pass
class InactiveWalletError(WalletError): pass
class InvalidPINError(WalletError): pass
class InvalidTransferAmountError(WalletError): pass
class InsufficientWalletBalanceError(WalletError): pass
class TransferLimitError(WalletError): pass

class DigitalWallet:
    TRANSFER_LIMIT = 10000

    def __init__(self, wallet_id, pin, balance=0, active=True):
        self.wallet_id = wallet_id
        self.pin = str(pin)
        self.balance = float(balance)
        self.active = active

    def transfer_to(self, receiver, pin, amount):
        try:
            if not self.wallet_id or not receiver.wallet_id:
                raise InvalidWalletError("Wallet ID is required.")
            if not self.active or not receiver.active:
                raise InactiveWalletError("Both wallets must be active.")
            if str(pin) != self.pin:
                raise InvalidPINError("Incorrect PIN.")
            amount = float(amount)
            if amount <= 0:
                raise InvalidTransferAmountError("Transfer amount must be positive.")
            if amount > self.TRANSFER_LIMIT:
                raise TransferLimitError("Transfer exceeds the per-transaction limit.")
            if amount > self.balance:
                raise InsufficientWalletBalanceError("Insufficient wallet balance.")
        except (WalletError, ValueError, TypeError) as error:
            print("Transfer failed:", error)
            return False
        else:
            # Both balances are changed only after all checks pass.
            self.balance -= amount
            receiver.balance += amount
            print(f"Transferred Rs. {amount:.2f} successfully.")
            return True

wallet_a = DigitalWallet("W101", "1234", 8000)
wallet_b = DigitalWallet("W102", "5678", 1000)
wallet_a.transfer_to(wallet_b, "1234", 2000)
wallet_a.transfer_to(wallet_b, "1234", 20000)
print("Sender balance:", wallet_a.balance, "| Receiver balance:", wallet_b.balance)
```

The application should continue running after recoverable errors.

```python
# Menu-driven retail inventory and sales application
class InventoryError(Exception): pass
class ProductMissingError(InventoryError): pass
class InvalidStockQuantityError(InventoryError): pass
class InsufficientInventoryError(InventoryError): pass

class RetailInventory:
    def __init__(self):
        self.items = {}

    def add_stock(self, name, quantity):
        try:
            quantity = int(quantity)
            if not name:
                raise ValueError("Product name is required.")
            if quantity <= 0:
                raise InvalidStockQuantityError("Quantity must be positive.")
        except (ValueError, TypeError, InvalidStockQuantityError) as error:
            print("Could not add stock:", error)
            return
        self.items[name] = self.items.get(name, 0) + quantity
        print(f"Added {quantity} unit(s) of {name}.")

    def view_inventory(self):
        print("Inventory:", self.items if self.items else "(empty)")

    def search_product(self, name):
        try:
            if name not in self.items:
                raise ProductMissingError(f"{name!r} not found.")
        except ProductMissingError as error:
            print(error)
            return
        print(f"{name}: {self.items[name]} unit(s)")

    def sell_product(self, name, quantity):
        try:
            if name not in self.items:
                raise ProductMissingError(f"{name!r} not found.")
            quantity = int(quantity)
            if quantity <= 0:
                raise InvalidStockQuantityError("Sale quantity must be positive.")
            if quantity > self.items[name]:
                raise InsufficientInventoryError("Not enough stock.")
        except ProductMissingError as error:
            print(error)
        except (ValueError, TypeError, InvalidStockQuantityError, InsufficientInventoryError) as error:
            print("Sale failed:", error)
        else:
            self.items[name] -= quantity
            print(f"Sold {quantity} unit(s) of {name}.")
        finally:
            print("Inventory operation finished.")

store = RetailInventory()
store.add_stock("Bottle", 20)
store.add_stock("Lunch Box", 12)
store.view_inventory()
store.search_product("Bottle")
store.sell_product("Bottle", 3)
store.sell_product("Bottle", 100)
store.view_inventory()
```

## ### Food Delivery Application

### Goal:
Create a complete console-based food delivery application using exception handling.

### Steps:
- Register customers.
- Select a restaurant.
- Display menu items.
- Take item quantities.
- Validate stock.
- Validate minimum order amount.
- Validate delivery area.
- Process payment.
- Create custom exception classes for business-rule failures.
- Use `try`, specific `except` blocks, `else`, `finally`, and `raise`.
- Prevent invalid operations from partially changing important application data.

### Example:
Customer: Ali
Restaurant: ABC Foods
Item: Burger
Quantity: 2
→ Order placed successfully

```python
# Console-style food delivery application with exception handling
class FoodOrderError(Exception): pass
class CustomerNotFoundError(FoodOrderError): pass
class RestaurantNotFoundError(FoodOrderError): pass
class ItemUnavailableError(FoodOrderError): pass
class InvalidOrderQuantityError(FoodOrderError): pass
class MinimumOrderError(FoodOrderError): pass
class DeliveryAreaError(FoodOrderError): pass
class PaymentError(FoodOrderError): pass

customers = {"C1": "Ali", "C2": "Sara"}
restaurants = {
    "ABC Foods": {
        "menu": {"Burger": {"price": 500, "stock": 10}, "Fries": {"price": 200, "stock": 20}},
        "minimum": 500,
        "areas": ["Nazimabad", "North Nazimabad"]
    }
}
orders = []

def place_food_order(customer_id, restaurant_name, items, area, payment):
    try:
        if customer_id not in customers:
            raise CustomerNotFoundError("Customer is not registered.")
        if restaurant_name not in restaurants:
            raise RestaurantNotFoundError("Restaurant not found.")
        restaurant = restaurants[restaurant_name]
        if area not in restaurant["areas"]:
            raise DeliveryAreaError("Delivery is not available in this area.")

        total = 0
        validated_items = []
        for item, quantity_value in items.items():
            if item not in restaurant["menu"]:
                raise ItemUnavailableError(f"{item} is not on the menu.")
            quantity = int(quantity_value)
            if quantity <= 0:
                raise InvalidOrderQuantityError("Quantity must be positive.")
            details = restaurant["menu"][item]
            if quantity > details["stock"]:
                raise ItemUnavailableError(f"Not enough stock for {item}.")
            total += details["price"] * quantity
            validated_items.append((item, quantity))

        if total < restaurant["minimum"]:
            raise MinimumOrderError(f"Minimum order is Rs. {restaurant['minimum']}.")
        if str(payment).lower() not in {"cash", "card", "wallet"}:
            raise PaymentError("Choose cash, card, or wallet.")
    except (FoodOrderError, ValueError, TypeError) as error:
        print("Order could not be placed:", error)
        return False
    else:
        # Commit stock and order only after all validations pass.
        for item, quantity in validated_items:
            restaurant["menu"][item]["stock"] -= quantity
        orders.append({
            "customer": customer_id, "restaurant": restaurant_name,
            "items": validated_items, "area": area, "total": total,
            "payment": str(payment).lower()
        })
        print(f"Order placed successfully for {customers[customer_id]}. Total: Rs. {total:.2f}")
        return True
    finally:
        print("Food delivery order process finished.")

place_food_order("C1", "ABC Foods", {"Burger": 2}, "Nazimabad", "cash")
place_food_order("C2", "ABC Foods", {"Burger": 100}, "Nazimabad", "wallet")
```
